Crear una decision de compra donde: El producto sea economicamente accesible - Que tenga presencia competitiva -Que tenga buenos resultados -Si tiene launcher le da valor -Que exista incertidumbre

```python
dataset["score_compra"] = 0
```

Calcular la capacidad economica donde se interpreta

≤ 8% del ingreso     → bastante accesible
8–12%                → compra posible
> 20%                → carga económica alta

```python
dataset.loc[
    dataset["precio_ingreso_pct"] <= 8,
    "score_compra"
] += 2

dataset.loc[
    (dataset["precio_ingreso_pct"] > 8) &
    (dataset["precio_ingreso_pct"] <= 12),
    "score_compra"
] += 1

dataset.loc[
    dataset["precio_ingreso_pct"] > 20,
    "score_compra"
] -= 2
```

```python
dataset.loc[
    dataset["meta_usage_pct"] >= 20,
    "score_compra"
] += 2

dataset.loc[
    (dataset["meta_usage_pct"] >= 5) &
    (dataset["meta_usage_pct"] < 20),
    "score_compra"
] += 1
```

```python
dataset.loc[
    dataset["meta_first_rate_pct"] >= 32,
    "score_compra"
] += 1
```

```python
dataset.loc[
    dataset["launcher_included"] == 1,
    "score_compra"
] += 0.5
```

Todo esto maneja las necesidades mencionadas y medidas agregando los valores necesarios para despues moldear con el PCA 

Esta parte solo creamos ruido ára el dataset basando como si una persona decidiera comprar o no

```python
np.random.seed(42)

dataset["ruido_decision"] = np.random.normal(
    loc=0,
    scale=0.75,
    size=len(dataset)
)
```

```python
dataset["score_final"] = (
    dataset["score_compra"]
    + dataset["ruido_decision"]
)
```

creamos la decision de comprar

```python
dataset["comprar"] = (
    dataset["score_final"] >= 2.5
).astype(int)
```

```python
dataset["comprar"].value_counts()
```

```python
dataset["comprar"].value_counts(normalize=True)
```

```python
plt.figure(figsize=(10,6))

for clase, color, etiqueta in [
    (0, "red", "No comprar"),
    (1, "green", "Comprar")
]:
    subset = dataset[dataset["comprar"] == clase]

    plt.scatter(
        subset["precio_ingreso_pct"],
        subset["meta_usage_pct"],
        alpha=0.7,
        label=etiqueta,
        color=color
    )

plt.xlabel("Precio como % del ingreso")
plt.ylabel("Usage competitivo (%)")
plt.title("Decisión de compra: costo económico vs meta")
plt.legend()

plt.show()
```

```python
resumen_productos = (
    dataset
    .groupby(["product_code", "product_name"])
    .agg(
        escenarios=("comprar", "size"),
        compras=("comprar", "sum"),
        porcentaje_compra=("comprar", "mean"),
        precio=("price_median_mxn", "first"),
        meta_usage=("meta_usage_pct", "first")
    )
    .reset_index()
)

resumen_productos["porcentaje_compra"] *= 100

display(resumen_productos)
```
