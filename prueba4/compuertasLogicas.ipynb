```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import matplotlib.pyplot as plt

# Datos de entrada y salida (compuerta XOR)
training_data = np.array([[0,0],[0,1],[1,0],[1,1]], "float32")
target_data = np.array([[0],[1],[1],[0]], "float32")

# Definición de la arquitectura del modelo
model = Sequential()
model.add(Dense(16, input_dim=2, activation='relu'))
model.add(Dense(1, activation='sigmoid'))

# Compilación del modelo
model.compile(loss='mean_squared_error',
              optimizer='adam',
              metrics=['binary_accuracy'])

# Entrenamiento del modelo
history = model.fit(training_data, target_data, epochs=1000)

# Evaluación del modelo y predicciones
scores = model.evaluate(training_data, target_data)
print("\n%s: %.2f%%" % (model.metrics_names[1], scores[1]*100))
print(model.predict(training_data).round())

# Gráfica de precisión y pérdida
plt.plot(history.history['binary_accuracy'])
plt.plot(history.history['loss'])
plt.title('Precisión del modelo')
plt.ylabel('Precisión')
plt.xlabel('Época')
plt.legend(['Entrenamiento', 'Test'], loc='upper left')
plt.show()
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt

# Datos de entrada y salida (compuerta OR)
training_data = np.array([[0,0],[0,1],[1,0],[1,1]], "float32")
target_data = np.array([[0],[1],[1],[1]], "float32") 


model = Sequential()
model.add(Dense(1, input_dim=2, activation='sigmoid'))

# Compilación del modelo
model.compile(loss='binary_crossentropy',
              optimizer=Adam(learning_rate=0.1),
              metrics=['binary_accuracy'])

# Entrenamiento del modelo
history = model.fit(training_data, target_data, epochs=1000, verbose=0)

# Evaluación del modelo y predicciones
scores = model.evaluate(training_data, target_data)
print("\n%s: %.2f%%" % (model.metrics_names[1], scores[1]*100))
print(model.predict(training_data).round())

# Gráfica de precisión y pérdida
plt.plot(history.history['binary_accuracy'])
plt.plot(history.history['loss'])
plt.title('Precisión del modelo (OR)')
plt.ylabel('Valor')
plt.xlabel('Época')
plt.legend(['Precisión', 'Pérdida (Loss)'], loc='center right')
plt.show()
```
